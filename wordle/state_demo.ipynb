```python
from wordle.state import *
from wordle.pattern import *

current_hist = tuple()
candidates_string = ["apple", "grape", "peach"]
candidates=range(len(candidates_string))
myState = GameState(candidates=candidates, history=current_hist)
def print_state(state: GameState):
    print("Current state of the game:")
    print("Guesses made:", state.guess_count)
    print("Is the game solved?", state.solved)
    print("Is the game failed?", state.failed)
    print("Is the game over?", state.done)
    print("Remaining possible answers:", state.remaining)
    print("History of guesses and patterns:")
    for guess, pattern in state.history:
        print(f"  Guess: {guess}, Pattern: {decode_pattern(pattern)}")
```

```python
print_state(myState)
```

```python
guess = "apple"
answer = "peach"
pattern = compute_pattern(guess, answer)
pattern_matrix = PatternMatrix(WordDistribution(candidates_string))
myState=myState.update(guess, pattern, pattern_matrix)
```

```python
print_state(myState)
```
