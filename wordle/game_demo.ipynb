```python
from wordle.game import *
from wordle.words import *
from wordle.pattern import *
import numpy
game = WordleGame.build()
```

```python
state,target=game.new_game(rng=np.random.default_rng(0))
print("Initial state: ", state)
print("The target word is: ", target)
```

```python
print("Guessing 'slate'...")
state,pattern,_=game.step(state,"slate",target)
print("Pattern obtained: ", decode_pattern(pattern))
print("State: ", state)
```

```python
remaining_words = sorted(game.words[i] for i in state.candidates)
print("Remaining candidates: ", remaining_words)
```

```python
print("Guessing 'angst'...")
state, pattern, _ = game.step(state, "angst", target)
print("Pattern obtained: ", decode_pattern(pattern))
print("State: ", state)
```
