```python
from words import *
rng = np.random.default_rng(42)  # fixed seed for reproducibility
distr = WordDistribution.default()  # default constructor with Zipf >= 1.0
print("Is 'apple' in the distribution?:", distr.contains("apple"))     # check if 'apple' is in the list
print("Probability of 'apple':", distr.probability("apple"))  # probability of 'apple'
print("Sample word:", distr.sample(rng=rng))
print("Number of words in the distribution:", len(distr))

print("\nCreating a distribution from a bigger zipf with 3.0")
distr = WordDistribution.from_wordfreq(3.0)
print("Is 'apple' in the distribution?:", distr.contains("apple"))  # check if 'apple' is in the list
print("Sample word:", distr.sample(rng=rng))
print("Number of words in the distribution:", len(distr))


print("\nCreating a distribution from a custom list of words uniformly distributed")
distr = WordDistribution(["crane", "slate"])
print("Is 'crane' in the distribution?:", distr.contains("crane"))
print("Probability of 'crane':", distr.probability("crane"))
print("Sample word:", distr.sample(rng=rng))
print("Number of words in the distribution:", len(distr))
```
