```python
from wordle.pattern import *
from wordle.words import *
# Example usage
guess = "slate"
answer = "crane"
pattern = compute_pattern(guess, answer)
print(
    f"Pattern for guess '{guess}' and answer '{answer}': {decode_pattern(pattern)}"
)
dict = WordDistribution(["slate", "crane", "crank"])
pattern_matrix = PatternMatrix(dict)
print(f"Pattern matrix:\n{pattern_matrix._matrix}")
print(
    f"Pattern for guess 'slate' and answer 'crane' from pattern matrix: {decode_pattern(pattern_matrix.get('slate', 'crane'))}"
)
```
